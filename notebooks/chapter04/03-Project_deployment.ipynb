# 9. 项目部署
## 9.1. 本地部署并对接`LangSmith`
**`LangSmith`** 是 **`LangChain`** 官方提供的 **`Agent`** 应用工程平台，主要用于对基于 **`LangGraph`**、**`LangChain`** 等框架开发的 **`Agent`** 应用进行运行链路追踪、调试、评估和生产监控。

**`LangChain`** 官方提供了开箱即用的 **`LangSmith` 云服务**，访问地址为：
https://smith.langchain.com/

此外，**`LangSmith`** 也支持混合部署和私有化部署。

除了基本的运行链路追踪，**`LangSmith` 云服务** 还提供了用于调试的 **`studio`**，要使用这个模块的功能，需要在本地部署 **`LangGraph`** 项目，这正是本节要完成的任务。

本节将部署一个经典的 **`HITL`** 应用，并在 **`LangSmith` 云服务** 的 **`studio`** 中调试。

# 总结

本节主要涉及 **Agent Chat UI** 和 **LangSmith**。更准确地说，前者是开源前端应用，后者是 Agent 开发、评估和运维平台；课程中用于可视化调试的 **LangSmith Studio** 是 LangSmith 提供的工具。


结合 LangGraph，可以这样区分各自的职责：

| 工具 | 主要职责 | 实际用途 |
| --- | --- | --- |
| **LangGraph** | 定义和执行节点、工具、状态与中断等业务流程 | 实现 Agent 的后端业务逻辑 |
| **Agent Chat UI** | 通过 Agent Server 连接后端，提供以 `messages` 为核心的聊天界面 | 展示流式回答、工具调用和中断交互，快速搭建演示或聊天产品原型 |
| **LangSmith Studio** | 可视化运行图、检查中间状态、交互调试执行过程 | 开发时观察节点路径、调试中断与恢复；可连接本地或已部署的 Agent Server |
| **LangSmith 的追踪与评估功能** | 记录执行链路、评估效果、监控线上表现 | 分析模型输入输出、工具调用、报错、耗时和 token 使用情况，对比修改前后的质量 |

以“查询天气前需要人工审批”为例：**LangGraph 实现审批逻辑，Agent Chat UI 提供用户操作界面，Studio 帮助开发者调试流程，LangSmith 追踪帮助分析实际执行记录。**

实际开发中的使用建议：

- **快速展示或交付内部助手**：使用 Agent Chat UI，省去从零编写聊天前端的工作；正式产品可以按业务需要定制界面、登录与权限。
- **开发复杂流程**：使用 Studio 检查状态、节点路径和中断恢复行为。
- **测试与上线后优化**：使用 LangSmith 追踪问题，通过测试数据集评估修改效果，并观察线上延迟、成本和质量。

这些工具可以按需选用，LangGraph 应用不要求全部接入。当前学习阶段，优先熟悉 **Studio 的状态检查和中断恢复**，再阅读一次完整的运行追踪；需要让别人体验聊天应用时，再接入 Agent Chat UI。

参考：[Agent Chat UI 官方文档](https://docs.langchain.com/oss/python/langchain/ui)、[LangSmith Studio 官方文档](https://docs.langchain.com/langsmith/studio)、[LangSmith 官方介绍](https://www.langchain.com/resources/what-is-langsmith)。
